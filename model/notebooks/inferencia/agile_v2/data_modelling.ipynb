## China Software Effort Estimation Dataset

**Origem:** Fang Hon Yun (2010)  
**Linhas:** ~499 projetos  
**Domínio:** Projetos de desenvolvimento de software corporativo  
**Objetivo:** Estimar esforço de desenvolvimento com base em Function Points  

Este dataset é amplamente utilizado em pesquisas de **Software Effort Estimation** e contém métricas baseadas em **Function Points** para representar o tamanho funcional do software.

### Variáveis principais

| Coluna | Descrição |
|------|-----------|
| `AFP` | Adjusted Function Points – tamanho funcional ajustado do software |
| `Input` | Número de entradas externas do sistema |
| `Output` | Número de saídas externas |
| `Inquiry` | Número de consultas externas |
| `File` | Número de arquivos lógicos internos |
| `Interface` | Número de arquivos de interface externa |
| `Added` | Quantidade de funcionalidades adicionadas |
| `Changed` | Quantidade de funcionalidades modificadas |
| `Deleted` | Quantidade de funcionalidades removidas |
| `Effort` | Esforço real do projeto (person-hours) |

agora aqui 

### Interpretação das métricas

Os atributos `Input`, `Output`, `Inquiry`, `File` e `Interface` representam componentes utilizados no cálculo de **Function Points**, que medem o tamanho funcional do sistema independentemente da tecnologia utilizada.

### Variável alvo para modelagem

Para tarefas de **Machine Learning**, a variável geralmente utilizada como alvo é **"Effort"** representando o esforço total necessário para desenvolver o projeto.

### Possíveis features derivadas

Algumas métricas úteis que podem ser calculadas:

- **Produtividade**
    Productivity = Effort / AFP
 
- **Mudança funcional**
    Change_Rate = (Added + Changed + Deleted) / AFP




```python
url = "https://raw.githubusercontent.com/Derek-Jones/Software-estimation-datasets/main/china.arff"

response = urllib.request.urlopen(url)
text = response.read().decode("utf-8")

data, meta = arff.loadarff(io.StringIO(text))
china = pd.DataFrame(data)

china = china.applymap(lambda x: x.decode() if isinstance(x, bytes) else x)
china
```

## Desharnais Software Effort Estimation Dataset

**Origem:** Jean-Marc Desharnais (1988)  
**Projetos:** 80  
**Domínio:** Projetos comerciais de desenvolvimento de software  
**Objetivo:** Estimar esforço de desenvolvimento com base em Function Points e características da equipe.

Este dataset contém projetos reais analisados em um estudo estatístico sobre produtividade de equipes de desenvolvimento.

### Variáveis principais

| Coluna | O que o valor representa | Unidade / Tipo |
|------|---------------------------|---------------|
| `Project` | Identificador único do projeto | ID numérico |
| `TeamExp` | Experiência média da equipe de desenvolvimento | Anos |
| `ManagerExp` | Experiência do gerente do projeto | Anos |
| `YearEnd` | Ano em que o projeto foi concluído | Ano |
| `Length` | Duração total do projeto | Meses |
| `Effort` | Esforço total gasto no projeto | Person-hours |
| `Transactions` | Número de transações funcionais do sistema | Contagem |
| `Entities` | Número de entidades de dados do sistema | Contagem |
| `PointsNonAdjust` | Function Points não ajustados (UFP – Unadjusted Function Points) | Function Points |
| `Adjustments` | Fator de ajuste baseado em características gerais do sistema | Fator numérico |
| `PointsAdjust` | Function Points ajustados (AFP – Adjusted Function Points) | Function Points |
| `Language` | Linguagem de programação principal utilizada | Código categórico |

#### Language — Linguagem de programação

| Código | Significado |
|------|-------------|
| 1 | Basic Cobol |
| 2 | Advanced Cobol |
| 3 | 4GL |

---

### Variável alvo

**Effort**

Representa o esforço total necessário para desenvolver o sistema.

### Observações

- `PointsAdjust` é normalmente a melhor métrica de **tamanho do software**.
- `TeamExp` e `ManagerExp` permitem estudar **influência do fator humano** na produtividade.

## NASA93 Software Effort Estimation Dataset

**Origem:** NASA / COCOMO research dataset  
**Projetos:** 93  
**Domínio:** Projetos aeroespaciais e sistemas críticos  
**Objetivo:** Modelar esforço de desenvolvimento com base em fatores do modelo COCOMO.

Este dataset contém projetos utilizados para validar e calibrar modelos de estimativa de esforço baseados em **COCOMO (Constructive Cost Model)**.

### Variáveis principais

| Coluna | Descrição |
|------|-----------|
| `rely` | Confiabilidade requerida do sistema |
| `data` | Tamanho do banco de dados |
| `cplx` | Complexidade do produto |
| `time` | Restrições de tempo de execução |
| `stor` | Restrições de memória |
| `virt` | Volatilidade da máquina virtual |
| `turn` | Tempo de turnaround |
| `acap` | Capacidade dos analistas |
| `pcap` | Capacidade dos programadores |
| `aexp` | Experiência na aplicação |
| `lexp` | Experiência na linguagem |
| `modp` | Uso de práticas modernas |
| `tool` | Uso de ferramentas de desenvolvimento |
| `sced` | Compressão do cronograma |
| `kloc` | Tamanho do sistema (milhares de linhas de código) |
| `effort` | Esforço total do projeto (person-months) |
| `defects` | Número de defeitos |
| `months` | Duração do projeto |

### Escala usada nos fatores COCOMO

| Código | Significado |
|------|-------------|
| vl | Very Low (Muito baixo) |
| l | Low (Baixo) |
| n | Nominal (Normal) |
| h | High (Alto) |
| vh | Very High (Muito alto) |
| xh | Extra High (Extremamente alto) |

### Variável alvo
- `effort`: Esforço total do projeto, medido em person-months. Esta é a variável que queremos prever com base nas outras características do projeto.
### Observações

- `kloc` representa o **tamanho do software**.
- As variáveis de confiabilidade, complexidade e capacidade da equipe são **drivers do modelo COCOMO**
